```python
# now just iterate it, but how many times. this is hte hyperparameter that we elarn 
# larger steps will take longer will the vocab size and shorter teh sequence. 
# we need to fidn a sweet spot.
# GPT uses 1024 tokens


def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

def merge(ids, pair, idx):
  newids = []
  i = 0
  while i < len(ids):
    if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
      newids.append(idx)
      i += 2
    else:
      newids.append(ids[i])
      i += 1
  return newids

# ---
vocab_size = 276 # the desired final vocabulary size
num_merges = vocab_size - 256
ids = list(tokens) # copy so we don't destroy the original list

merges = {} # (int, int) -> int
for i in range(num_merges):
  stats = get_stats(ids)
  pair = max(stats, key=stats.get)
  idx = 256 + i
  print(f"merging {pair} into a new token {idx}")
  ids = merge(ids, pair, idx)
  merges[pair] = idx
```

```python
def encode(text):
    # get the raw bytes
    tokens = list(text.encode("utf-8"))
    # we need to do top-to-bottom merges
    while len(tokens) > 2:
        # find pairs of bytes that are consective and mergable
        stats = get_stats(tokens)
        #Min takes a key which gives us the function that is
        # going to return a value over which we're going to do the Min and the one we care about is we're we care about taking the minimum of the stats
        # and the stats is a dictionary where the keys are the pairs and the values are the counts
        # so we're going to take the minimum of the counts
        pair = min(stats, key=lambda p: merges.get(p, float('inf')))
        if pair not in merges:
            break # no more merges left
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
    return tokens
print(encode("Hello, world!"))
```

# Optimizations and Alternative Techniques for BPE Tokenization

The following cells explore various ways to optimize the BPE implementation and alternative tokenization techniques.

## 1. Optimized get_stats() using collections.Counter

More Pythonic and slightly faster for large inputs due to C-optimized internals.

```python
from collections import Counter

def get_stats_counter(ids):
    """Optimized stats using Counter - more Pythonic and C-optimized."""
    return Counter(zip(ids, ids[1:]))

# Test it
stats_counter = get_stats_counter(tokens)
print("Top 5 pairs:", stats_counter.most_common(5))
```

## 2. Priority Queue / Heap for Finding Max Pair

Instead of max() on the entire dict each iteration O(vocab), use a heap for O(log vocab).

```python
import heapq

def train_bpe_with_heap(tokens, num_merges):
    """
    BPE training using a max-heap for faster pair selection.
    Uses lazy deletion - invalid entries are skipped when popped.
    """
    ids = list(tokens)
    merges = {}
    stats = get_stats_counter(ids)
    
    # Max-heap (negate counts for max behavior since heapq is min-heap)
    heap = [(-count, pair) for pair, count in stats.items()]
    heapq.heapify(heap)
    
    for i in range(num_merges):
        # Find valid max pair (lazy deletion)
        while heap:
            neg_count, pair = heapq.heappop(heap)
            # Verify count is still valid (may be stale after merges)
            if stats.get(pair, 0) == -neg_count:
                break
        else:
            break  # No more valid pairs
        
        idx = 256 + i
        print(f"merging {pair} into a new token {idx}")
        ids = merge(ids, pair, idx)
        merges[pair] = idx
        
        # Recompute stats and push new pairs to heap
        stats = get_stats_counter(ids)
        for p, c in stats.items():
            heapq.heappush(heap, (-c, p))
    
    return ids, merges

# Example usage (smaller num_merges for demo)
# ids_heap, merges_heap = train_bpe_with_heap(tokens, 5)
```

## 3. Trie-based Encoding for Faster Token Lookup

Instead of iteratively merging during encoding, use a Trie to find longest matching tokens in O(n) time.

```python
class TrieNode:
    """Node in a Trie for fast token lookup."""
    def __init__(self):
        self.children = {}
        self.token_id = None

def build_vocab_trie(vocab):
    """Build a Trie from vocabulary for O(1) prefix matching."""
    root = TrieNode()
    for idx, token_bytes in vocab.items():
        node = root
        for byte in token_bytes:
            if byte not in node.children:
                node.children[byte] = TrieNode()
            node = node.children[byte]
        node.token_id = idx
    return root

def encode_with_trie(text, trie_root):
    """
    Encode text using Trie-based longest match.
    Time complexity: O(n) where n is input length.
    """
    tokens = []
    data = text.encode("utf-8")
    i = 0
    while i < len(data):
        node = trie_root
        last_match = (i + 1, data[i])  # fallback: single byte
        j = i
        # Find longest matching token
        while j < len(data) and data[j] in node.children:
            node = node.children[data[j]]
            j += 1
            if node.token_id is not None:
                last_match = (j, node.token_id)
        tokens.append(last_match[1])
        i = last_match[0]
    return tokens

# Build trie from our vocab
trie_root = build_vocab_trie(vocab)

# Test encoding
encoded_trie = encode_with_trie("Hello, world!", trie_root)
print("Trie-encoded:", encoded_trie)
print("Original encode:", encode("Hello, world!"))
print("Decoded:", decode(encoded_trie))
```

## 4. Parallelization with Multiprocessing

For large corpora, parallelize stats collection across chunks of data.

```python
from multiprocessing import Pool
from functools import reduce

def merge_stats(stats_list):
    """Merge multiple stats dictionaries."""
    return reduce(
        lambda a, b: {k: a.get(k, 0) + b.get(k, 0) for k in set(a) | set(b)},
        stats_list
    )

def parallel_get_stats(token_chunks, num_workers=4):
    """
    Compute pair statistics in parallel across chunks.
    Useful for very large corpora.
    """
    with Pool(num_workers) as pool:
        stats_list = pool.map(get_stats, token_chunks)
    return merge_stats(stats_list)

# Example: split tokens into chunks
def chunk_list(lst, n_chunks):
    """Split list into n roughly equal chunks."""
    chunk_size = len(lst) // n_chunks + 1
    return [lst[i:i + chunk_size] for i in range(0, len(lst), chunk_size)]

# Demo (commented out to avoid multiprocessing issues in notebook)
# chunks = chunk_list(tokens, 4)
# parallel_stats = parallel_get_stats(chunks)
# print("Parallel stats computed for", len(parallel_stats), "pairs")
```

## 5. Benchmarking Original vs Optimized

Compare performance of different implementations.

```python
import time

def benchmark(func, *args, runs=5):
    """Simple benchmark helper."""
    times = []
    for _ in range(runs):
        start = time.perf_counter()
        result = func(*args)
        times.append(time.perf_counter() - start)
    return min(times), result

# Benchmark get_stats implementations
t1, _ = benchmark(get_stats, tokens)
t2, _ = benchmark(get_stats_counter, tokens)
print(f"get_stats (dict):    {t1*1000:.3f} ms")
print(f"get_stats (Counter): {t2*1000:.3f} ms")
print(f"Speedup: {t1/t2:.2f}x")

# Benchmark encoding implementations  
test_text = "The quick brown fox jumps over the lazy dog. " * 100
t3, enc1 = benchmark(encode, test_text)
t4, enc2 = benchmark(encode_with_trie, test_text, trie_root)
print(f"\nencode (iterative merge): {t3*1000:.3f} ms")
print(f"encode (trie):            {t4*1000:.3f} ms")
print(f"Speedup: {t3/t4:.2f}x")
print(f"Results match: {enc1 == enc2}")
```

## 6. Alternative Tokenization Algorithms

### WordPiece (used by BERT)
Similar to BPE but uses likelihood maximization instead of frequency.

### Unigram Language Model (used by SentencePiece, T5)
Probabilistic model that starts with large vocab and prunes iteratively.

```python
# Simplified WordPiece-style scoring
# WordPiece uses: score(pair) = freq(pair) / (freq(first) * freq(second))
# This favors pairs where both tokens commonly appear together

def get_wordpiece_scores(ids):
    """
    WordPiece-style scoring: likelihood-based instead of frequency-based.
    score = freq(pair) / (freq(token1) * freq(token2))
    """
    # Get pair frequencies
    pair_freq = Counter(zip(ids, ids[1:]))
    
    # Get individual token frequencies
    token_freq = Counter(ids)
    
    # Calculate WordPiece scores
    scores = {}
    for pair, freq in pair_freq.items():
        t1, t2 = pair
        # Avoid division by zero
        score = freq / (token_freq[t1] * token_freq[t2])
        scores[pair] = score
    
    return scores

# Compare BPE vs WordPiece pair selection
bpe_stats = get_stats_counter(tokens)
wp_scores = get_wordpiece_scores(tokens)

print("Top 5 BPE pairs (by frequency):")
for pair, count in bpe_stats.most_common(5):
    print(f"  {pair}: count={count}")

print("\nTop 5 WordPiece pairs (by likelihood score):")
wp_sorted = sorted(wp_scores.items(), key=lambda x: x[1], reverse=True)[:5]
for pair, score in wp_sorted:
    print(f"  {pair}: score={score:.6f}, count={bpe_stats[pair]}")
```

# Sentencepiece
### sentencepiece

Commonly used because (unlike tiktoken) it can efficiently both train and inference BPE tokenizers. It is used in both Llama and Mistral series.

[sentencepiece on Github link](https://github.com/google/sentencepiece).

**The big difference**: sentencepiece runs BPE on the Unicode code points directly! It then has an option `character_coverage` for what to do with very very rare codepoints that appear very few times, and it either maps them onto an UNK token, or if `byte_fallback` is turned on, it encodes them with utf-8 and then encodes the raw bytes instead.

TLDR:

- tiktoken encodes to utf-8 and then BPEs bytes
- sentencepiece BPEs the code points and optionally falls back to utf-8 bytes for rare code points (rarity is determined by character_coverage hyperparameter), which then get translated to byte tokens.
